**Linear Regression**

```python
binary = ['mainroad','guestroom','basement','hotwaterheating','airconditioning','parking','prefarea']
for col in binary:
    df[col] = df[col].map({'yes': 1 ,'no':0})
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
X = df[['area','bedrooms','bathrooms','stories','mainroad','guestroom','basement','hotwaterheating','airconditioning','prefarea']]
y = df[['price']]
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=2)
lr = LinearRegression()
lr.fit(X_train,y_train)
m= lr.predict(X_test)
print(m)
```

```python
from sklearn.metrics import mean_absolute_error
mae = mean_absolute_error(y_test,m)
print(mae)
```

```python
from sklearn.metrics import mean_squared_error
mae = mean_squared_error(y_test,m)
print(mae)
```

```python
from sklearn.metrics import root_mean_squared_error
mae = root_mean_squared_error(y_test,m)
print(mae)
```

```python
from sklearn.metrics import r2_score
mae = r2_score(y_test,m)
print(mae)
```

Observation
- The mean absolute error is 847506.653
- The mean squared error is 1325980290430.02
- The root mean squared error is 1151512.175
- The R2 score is 0.6263
